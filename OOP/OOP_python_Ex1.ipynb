```python
class Car:
  def __init__ (self,license_plate ,brand):
    self._brand = brand
    self.__license_plate = license_plate

  @property
  def brand(self):
    return self.__brand

  @brand.setter
  def brand(self, brand):
    self.__brand = brand

  @brand.setter
  def brand(self, brand):
    self.__brand = brand

  def drive (self):
    print(f"{self.brand} is brand your car")

  def __repr__(self):
    masked = "**/ *** " + self.__license_plate[-2:]
    return masked

```

# polymorphism
- یک متد یا مفهوم با نام یکسان می تواند بسته به نوع کلاس رفتار متفاوتی داشته باشد

```python
class Vehicle:
    def move(self, dest):
        print(f"{self.name} به {dest} می‌ره")


class Car(Vehicle):
    def move(self, dest):
        print(f"ماشین با بنزین به {dest} می‌ره")


class Bicycle(Vehicle):
    def move(self, dest):
        print(f"دوچرخه با رکاب به {dest} می‌ره")


class Airplane(Vehicle):
    def move(self, dest):
        print(f"هواپیما در آسمون به {dest} پرواز می‌کنه")
```

```python
vehicles = [Car(), Bicycle(), Airplane()]

for v in vehicles:
    v.move("تهران")     # ← این خط، polymorphism هست
```
