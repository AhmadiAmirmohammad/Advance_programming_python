کلاس فرزندان باید بتوانند جایگزین اشیاء کلاس والد شوند 

## 3. problem

```python
class Bird :
    def fly():
        print("flying")

class Pigeon(Bird):
    pass

class Panguin(Bird):
    def fly():
        raise Exception ("not flying")

def make_it_fly (bird: Bird):
    bird.fly()

make_it_fly(Panguin)
make_it_fly(Pigeon)  
```

## 3.solution

```python
class Bird :
    def move():
        print("moving")

class FlyBird(Bird):
    def fly():
        print("flying")

class Pigeon(FlyBird):
    pass
    
class Panguin(Bird):
    pass

def make_it_fly (bird_fly: FlyBird):
    bird_fly.fly()
make_it_fly(Pigeon)  
```

```python
def make_it_move(bird: Bird):
    bird.move()
make_it_move(Panguin)
```
