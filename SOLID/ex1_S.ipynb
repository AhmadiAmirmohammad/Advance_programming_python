## 1. problem (single responsibility)

```python
class Soldier:
    def __init__ (self, name , id_number, year_experience):
        self._name = name
        self.__id_number = id_number
        self.year_experience = year_experience

    def soldierReponsitory(self):
        print(f"successfully save Soldier name:{self._name} id: {self.__id_number} with experience {self.year_experience}")
    def calculateSalary (self):
        print(f"Salary is: {self.year_experience *1.4}")
```

## 1.Solution (Single responsibility)

```python
class Soldier: 
    def __init__(self, name, id_number, year_experience):
        self._name = name
        self.__id_number= id_number
        self.year_experience = year_experience

    @property
    def id_number(self):
        return self.__id_number

    @property 
    def name(self):
        return self._name

class SoldierReponsitory: 
    def __init__(self, soldier):
        self. soldier = soldier        
    def register(self):
        print(f"successfully save Soldier name:{self.soldier.name} id: {self.soldier.id_number} with experience {self.soldier.year_experience}")
```

```python
AmirAliAhmadi = Soldier("AmirAliAhmadi", "1464", 12)
```

```python
soldierReponsitory= SoldierReponsitory(AmirAliAhmadi)
soldierReponsitory.register()
```
