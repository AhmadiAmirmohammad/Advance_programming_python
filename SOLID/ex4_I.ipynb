# 4. Interface segrigation 
- جداکردن اینترفیس ها
- بجای داشتن یک اینترفیس بزرگ چندتا اینترفیس کوچک و اختصاصی
- که چی بشه ؟ هر کلاس فقط باید متدهایی را که لازم دارد پیاده سازی کند

```python
from abc import ABC, abstractmethod
class PrinterMachine(ABC):
    @abstractmethod
    def printMethod():
        print("printing ")

    @abstractmethod
    def faxMethod():
        print("faxing ")

    @abstractmethod
    def scanMethod():
        print("scanning")

class AllInOnePrinter(PrinterMachine):
    pass

class SimplePrinter(PrinterMachine):
    def faxMethod():
        raise Exception("can not faxing")

def printer_fax (printer:PrinterMachine):
    printer.faxMethod()

printer_fax(SimplePrinter)
```
