## 2. problem

```python
class CalculateDiscount: 
    def __init__(self, user_role , perice):
        self.__user_role= user_role
        self.perice = perice
        
    def calculate(self):
        if self.__user_role == "general": 
             print(f"perice is {self.perice}")
        if self.__user_role == "vip": 
             print  ( f" perice is: {self.perice * 0.5}")
        else: return 

```

```python
calculateDiscount  = CalculateDiscount("general",50)
calculateDiscount.calculate()
```

مساله ساز وقتی میشه که می خواهیم یک نوع کد تخفیف دیگه اضافه کنیم 

## 2. solution

```python
from abc import ABC, abstractmethod

class DiscountStrategy(ABC):
    @abstractmethod
    def calculate(self, perice):
        pass
```

```python
class GeneralDiscount(DiscountStrategy):
    def calculate ( self, perice):
        print(f"perice is {perice}")

class VipDiscount(DiscountStrategy):
    def calculate (self, perice):
        print( f"perice is: {perice * 0.5}")
```

```python
calculate_normal_discount = GeneralDiscount().calculate(50)
```

```python
class PericeCalculator():
    def final_price(self, price, discount: DiscountStrategy):
        return discount.calculate(price)     # نمی‌داند کدام تخفیف است
```

```python
calc = PericeCalculator()

calc.final_price(1000, GeneralDiscount())   
calc.final_price(1000, VipDiscount())  
```
