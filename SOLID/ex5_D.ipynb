# 5. Dipendency Inversion 
وابستگی معکوس

- ماژول های سطح بالا نباید به ماژول های سطح پایین وابستگی داشتند یا همان پیاده سازی خاص
- یعنی هر 2 باید به انتزاع یا همان اینترفیس وابسته باشند

```python
from abc import ABC , abstractmethod

# مرحله ۱: انتزاع (قرارداد)
class Database (ABC):

    @abstractmethod
    def save(self):
        pass
        
# مرحله ۲: پیاده‌سازی‌ها (سطح پایین)
class MySQLDatabase(Database):
    def save(self, data): print(f"ذخیره در MySQL: {data}")

class MongoDatabase(Database):
    def save(self, data): print(f"ذخیره در MongoDB: {data}")
        
#    مرحله ۳: کلاس سطح بالا (فقط به انتزاع وابسته)       
class UserService:
    def __init__(self, db: Database):    # ✅ از بیرون تزریق می‌شود
        self.db = db

    def register(self, user):
        self.db.save(user)    

#مرحله ۴: استفاده
service1 = UserService(MySQLDatabase())
service1.register("ali")        # ذخیره در MySQL: ali

service2 = UserService(MongoDatabase())
service2.register("sara")       # ذخیره در MongoDB: sara
```
